```python
# *****Project Name is Student Marks Analyzer*****

import numpy as np
marks=np.array([[75,82,87],
               [65,81,77],
               [60,58,90],
               [80,67,94]])
#1.Student marks average
avg_marks=np.average(marks)
print("Average Marks of Student",avg_marks)

#2.highest marks
max_marks=np.max(marks)
print("Highest marks ",max_marks)

#3.lowest marks
low_marks=np.min(marks)
print("lowest marks ",low_marks)

#4. find student who passed (avg>=60)
passed_students=np.where(avg_marks>=60)
print("student passed",passed_students)

#5.cout total passed and failed
print("Total passed:",len(passed_students))
print("Total failed:",len(avg_marks)-len(passed_students))
```
